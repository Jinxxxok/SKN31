09_입출력

프로그램 - 외부자원
인풋 아웃풋은 하나씩 존재

스트림이란? 값이 하나씩 이동하는 것을 스트림이라고함
인풋 스트림: 데이터를 입력받는 흐름
아웃풋 스트림: 데이터를 출력해주는 흐름

기본적인 스트림의 단위는 1byte씩 흘러간다고 보면됌

1bit = 기본 단위 한칸
1byte = 8bit

실제 데이터가 128byte라면 128번을 거쳐서 스트림한다고 보면된다.

중간에 순서를 바꿔서 읽어들일수는 없고 다시 읽어들이려면 스트림을 끊고 다시 시작해야함

IO 코딩 순서
파일 열기(연결) => 데이터를 파일에 읽거나 쓰기 => 파일 닫기(연결 끊기)

입력 연결은 입력밖에 못하고 출력 연결은 출력밖에 못한다.

[파일 열기] // open()
- 대상과 연결 한다.
open(file, mode='r', encoding=None)

file: 연결할 파일의 경로
mode: read / write (r / w), w 새로쓰기 모드
a 이어쓰기 모드, 
x 새로쓰기 모드 - 연결하려는 파일이 있으면 exception발생, 기존의 txt파일이 있으면 에러를 내버림
* 무조건 새로운 파일을 만들어서 사용할거야 라고 할때 사용
b binary모드, 
t Text모드 - text데이터 입출력시 사용

유니코드는 기본적으로 2byte

b: text파일을 제외하고 텍스트들은 전부 바이너리 모드

t: text 메모장으로 열수있는 파일은 t

데이터 종류의 default값은 t mode타입의 default값은 r

encoding: 
생략하면 os기본 encoding방식을 따른다.
window = cp949
Mac, Linux, Unix = Utf-8

os별로 인코딩 방식이 달라서 맞춰줘야 exception이 발생하지 않는다.


[데이터를 파일에 쓰기/읽기] // read() , write()

[파일 닫기] // close()

입력 메소드

readline = 한줄씩 읽어줌 끝이나면 공백을 가져옴

readlines = 한번에 다 읽은뒤 각각 줄을 리스트의 원소로 담아서 반환해준다

for in 문 사용 = ex) for line in fr: 하면 한줄 씩 넣어서 출력해준다.


'rt', 'wt': t모드 -> 입출력 타입: str read(): str, write('str')
'rb', 'wb': b모드 -> 입출력 타입: bytes read(): bytes, write(bytes)

bytes type: byte들을 모아 놓은 타입

t => str
b => 나머지 전부

b 변환이 어려운 상황 모든 클래스들마다 변환방식이 다르기때문에
pickle 모듈이 이를 대신 해줌.

문자열을 제외한 나머지 변환은 전부 pickle이 도맡아서 함.

객체 직렬화 -> 객체들의 속성값을 bytes로 변환해 출력하는 것
객체 역직렬화 -> bytes로 출력된 데이터를 읽어 다시 원래대로 객체화하는것

피클의 파일확장자는 보통 pkl or pickle 이라고 한다.
